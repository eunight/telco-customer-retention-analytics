```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# 1. Carga de datos crudos (usando rutas relativas por la estructura del proyecto)
df = pd.read_csv('../data/raw/telco.csv')

# 2. Eliminación de columnas que causan Fuga de Datos (Data Leakage)
columnas_a_eliminar = ['Churn Category', 'Churn Reason']
df = df.drop(columns=columnas_a_eliminar)

# 3. Corrección de tipos de datos encubiertos
# 'Total Charges' viene como texto porque los valores faltantes son espacios ' '
df['Total Charges'] = pd.to_numeric(df['Total Charges'], errors='coerce')

# 4. Imputación lógica de nulos
# Los nulos en 'Total Charges' corresponden a clientes en su primer mes (Tenure = 0). Su cargo total es 0.
df['Total Charges'] = df['Total Charges'].fillna(0)

# Los nulos en 'Internet Type' significan que el cliente no contrató internet.
df['Internet Type'] = df['Internet Type'].fillna('No')

# Los nulos en 'Offer' significan que el cliente no fue retenido con ninguna oferta especial.
df['Offer'] = df['Offer'].fillna('None')

# 5. Verificación final de integridad
print("Valores nulos restantes tras la limpieza:")
print(df.isnull().sum()[df.isnull().sum() > 0])
print(f"\nDimensiones de la matriz limpia: {df.shape}")
```
